# Q2) Conditional Density of the Multivariate Gaussian

Let $\mathbf{Z}$ be the concatenated random matrix
$\mathbf{Z} = \begin{bmatrix} \mathbf{X} \\ \mathbf{Y} \end{bmatrix} \in \mathbb{R}^{n+m}$.


The random variable $\mathbf{Z}$ is said to have a **joint multivariate normal distribution** if every linear combination of its components follows a univariate normal distribution.

For $\mathbf{Z}$ to possess a **joint density function** (meaning its distribution is absolutely continuous with respect to the Lebesgue measure on $\mathbb{R}^{n+m}$), its covariance matrix must be non-singular (positive definite).


The joint density function $f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z})$ for $\mathbf{z} \in \mathbb{R}^{n+m}$ is given by:

$$f_{\mathbf{X}, \mathbf{Y}}(\mathbf{z}) = \frac{1}{\sqrt{(2\pi)^{n+m} |\boldsymbol{\Sigma}|}} \exp\left( -\frac{1}{2} (\mathbf{z} - \boldsymbol{\mu})^T \boldsymbol{\Sigma}^{-1} (\mathbf{z} - \boldsymbol{\mu}) \right).$$

Here:

* **Mean ($\boldsymbol{\mu}$):** $\boldsymbol{\mu} = E[\mathbf{Z}] = \begin{bmatrix} \boldsymbol{\mu}_X \\ \boldsymbol{\mu}_Y \end{bmatrix} \in \mathbb{R}^{n+m}$, where $\boldsymbol{\mu}_X \in \mathbb{R}^n$ and $\boldsymbol{\mu}_Y \in \mathbb{R}^m$.
* **Covariance Matrix ($\boldsymbol{\Sigma}$):** A symmetric, positive definite matrix $\boldsymbol{\Sigma} \in \mathbb{R}^{(n+m) \times (n+m)}$ partitioned as:
$$\boldsymbol{\Sigma} = \begin{bmatrix} \boldsymbol{\Sigma}_{XX} & \boldsymbol{\Sigma}_{XY} \\ \boldsymbol{\Sigma}_{YX} & \boldsymbol{\Sigma}_{YY} \end{bmatrix}$$
    * $\boldsymbol{\Sigma}_{XX} = \text{Var}(\mathbf{X})$ (size $n \times n$)
    * $\boldsymbol{\Sigma}_{YY} = \text{Var}(\mathbf{Y})$ (size $m \times m$)
    * $\boldsymbol{\Sigma}_{XY} = \text{Cov}(\mathbf{X}, \mathbf{Y}) = \boldsymbol{\Sigma}_{YX}^T$ (size $n \times m$)
* **Determinant ($|\boldsymbol{\Sigma}|$):** The normalization factor ensuring the total integral is 1.

1. Show that the conditional distribution is given by $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y})$

$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\mathbf{S}_{XX}|}} \exp\left( -\frac{1}{2} (\mathbf{x} - \boldsymbol{\mu}^*)^T \boldsymbol{\Sigma}_{X\mid Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*) \right),$$
where
* $\boldsymbol{\Sigma}_{X\mid Y} = \boldsymbol{\Sigma}_{XX} - \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}\boldsymbol{\Sigma}_{YX}$
* $\boldsymbol{\mu}^* = \boldsymbol{\mu}_x + \boldsymbol{\Sigma}_{XY}\boldsymbol{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_y)$

$f_{X \mid Y}(x \mid y) = \frac{f_{X,Y}(x,y)}{f_Y(y)}$

The exponent of the joint density contains the quadratic form $Q_{\text{joint}}$. By using the block matrix inversion and the Schur complement of $\Sigma_{YY}$, we can complete the square and decompose $Q_{\text{joint}}$ into two distinct parts:

$$Q_{\text{joint}} = (y - \mu_Y)^T \Sigma_{YY}^{-1} (y - \mu_Y) + (x - \mu^*)^T S_{XX}^{-1} (x - \mu^*)$$

Where:

$S_{XX} = \Sigma_{XX} - \Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX}$ (This is the Schur complement of $\Sigma_{YY}$, which acts as the conditional covariance $\Sigma_{X|Y}$).

$\mu^* = \mu_X + \Sigma_{XY}\Sigma_{YY}^{-1}(y - \mu_Y)$.

$Q_{\text{marginal}} = (y - \mu_Y)^T \Sigma_{YY}^{-1} (y - \mu_Y)$

Dividing the joint density by the marginal density $\left( \frac{e^{-\frac{1}{2}Q_{\text{joint}}}}{e^{-\frac{1}{2}Q_{\text{marginal}}}} \right)$, subtracts the marginal exponent from the joint exponent ($Q_{\text{joint}} - Q_{\text{marginal}}$). The terms depending purely on $y$ cancel out perfectly

$Q_{\text{conditional}} = (x - \mu^*)^T S_{XX}^{-1} (x - \mu^*)$


Dividing the normalization constant of the joint density by the normalization constant of the marginal density:

Joint constant: $\frac{1}{\sqrt{(2\pi)^{n+m}|\boldsymbol{\Sigma}|}}$

Marginal constant: $\frac{1}{\sqrt{(2\pi)^m|\boldsymbol{\Sigma}_{YY}|}}$

This yields:

$$\frac{\frac{1}{\sqrt{(2\pi)^{n+m}|\Sigma|}}}{\frac{1}{\sqrt{(2\pi)^m|\Sigma_{YY}|}}} = \frac{\sqrt{(2\pi)^m|\Sigma_{YY}|}}{\sqrt{(2\pi)^{n+m}|\Sigma|}} = \frac{1}{\sqrt{(2\pi)^n \frac{|\Sigma|}{|\Sigma_{YY}|}}}$$

Simplifying using the determinant identity for block matrices:

$|\Sigma| = |\Sigma_{YY}| \cdot |S_{XX}|$

$\frac{|\Sigma|}{|\Sigma_{YY}|}=|S_{XX}|$

Combining the simplified normalization constant and the remaining quadratic form $Q_{\text{conditional}}$

$f_{\mathbf{X} \mid \mathbf{Y}}(\mathbf{x} \mid \mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n |\mathbf{S}_{XX}|}} \exp\left(-\frac{1}{2}(\mathbf{x} - \boldsymbol{\mu}^*)^T \mathbf{S}_{XX}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*)\right)$


# Q3) Continuous Fault Monitoring of a Spindle Motor

A factory monitors the health of a high-speed CNC spindle. The critical variable is the **Internal Temperature ($X$)**, but because the motor is sealed, they can only observe the **Total Spindle Power ($Y$)**.

**The Setup:**
1.  **Hidden Temperature ($X$):** The standardized temperature deviation follows a standard normal distribution:
    $$X \sim \mathscr{N}(0, 1)$$
2.  **Mechanical Load ($Q$):** The torque required to cut material creates "noise" in the power reading. It is independent of temperature and follows:
    $$Q \sim \mathscr{N}(0, 4)$$
3.  **The Measurement ($Y$):** The power sensor $Y$ measures the combined effect of thermal load and mechanical torque:
$$Y = X + Q$$

---

**Answer the followings:**



**1. Joint Distribution**
Identify the distribution of the random varaible $(X, Y)$.
* Show that $Y \sim \mathscr{N}(0, 5)$.
* Find the covariance $\text{Cov}(X, Y)$.

The mean of $Y$ is the sum of the means:
$E[Y] = E[X] + E[Q] = 0 + 0 = 0$

Because $X$ and $Q$ are independent, their variances add together:
$\text{Var}(Y) = \text{Var}(X) + \text{Var}(Q) = 1 + 4 = 5$

Thus, $Y \sim \mathcal{N}(0,5)$.

$\text{Cov}(X,Y) = \text{Cov}(X, X+Q) = \text{Cov}(X,X) + \text{Cov}(X,Q)$

Since $X$ and $Q$ are independent, $\text{Cov}(X,Q) = 0$. Therefore, $\text{Cov}(X,Y) = \text{Var}(X) = 1$.



**2. Information Geometry**
Explain why $X$ is **not** $\sigma(Y)$-measurable. In practical terms, if the power meter shows $Y = 5$, why can't the engineer know for certain if the motor is overheating ($X$ is high) or if the material being cut is simply very hard ($Q$ is high)?


By the Doob-Dynkin lemma, if a variable is $\sigma(Y)$-measurable, it means it must be purely a deterministic function of $Y$, written as $X = g(Y)$.

$X$ is not $\sigma(Y)$-measurable because observing $Y=5$ does not uniquely identify $X$. The value $Y=5$ could be caused by an overheating spindle ($X=5$) with no cutting noise ($Q=0$), or a completely normal spindle ($X=0$) cutting very hard material ($Q=5$). Because of the unobserved mechanical noise $Q$, $Y$ does not contain enough "information" to perfectly pinpoint the hidden temperature.

**3. Conditional Expectation (The Best Estimate)**
Compute the conditional expectation $E[X \mid Y]$.
* Using the property for joint normals: $E[X \mid Y = y] = \mu_X + \frac{\text{Cov}(X,Y)}{\text{Var}(Y)}(y - \mu_Y)$.
* Express this as a "signal-to-total-variance" ratio.


For a bivariate normal distribution,

$E[X \mid Y=y] = \mu_X + \frac{\text{Cov}(X,Y)}{\text{Var}(Y)}(y - \mu_Y)$

Substituting values ($\mu_X = 0$, $\mu_Y = 0$, $\text{Cov}(X,Y) = 1$, and $\text{Var}(Y) = 5$)

$E[X \mid Y] = 0 + \frac{1}{5}(Y - 0) = \frac{1}{5}Y$

The coefficient $\frac{1}{5}$ is exactly the ratio of the "signal" variance ($\text{Var}(X)=1$) over the total variance of the measurement ($\text{Var}(Y) = \text{Var}(X) + \text{Var}(Q) = 5$).

**4. Verification of the Residual**
Let $\widehat{X} = E[X \mid Y]$. Define the estimation error (residual) as $Z = X - \widehat{X}$.
* Prove that $E[Z] = 0$.
* Verify that the error $Z$ is uncorrelated with the observation $Y$ (the Orthogonality Principle).



$E[Z] = E[X - E[X \mid Y]] = E[X] - E[E[X \mid Y]]$


By the Law of Total Expectation (or Iterated Expectations), $E[E[X \mid Y]] = E[X]$.

 Therefore, $E[Z] = E[X] - E[X] = 0$.





</br>$E[ZY] = E[(X - E[X \mid Y])Y]$


By the Law of Iterated Expectations, this is $E[E[(X - E[X \mid Y])Y \mid Y]]$.

 Since $Y$ is $\sigma(Y)$-measurable, the Factorization Property allows us to pull it outside the inner expectation:$E[Y \cdot E[X - E[X \mid Y] \mid Y]]$


The inner term becomes $E[X \mid Y] - E[E[X \mid Y] \mid Y]$, which collapses to $E[X \mid Y] - E[X \mid Y] = 0$. Thus, the entire expectation is $0$.

Hence, $E[ZY]=0$

**5. Information Theory: Differential Entropy**
* Compute the differential entropy $h(X)$.
* Compute the conditional differential entropy $h(X \mid Y)$.
* *Note: The variance of the conditional distribution $(X \mid Y=y)$ is given by $\sigma_{X|Y}^2 = \sigma_X^2 - \frac{\text{Cov}(X,Y)^2}{\text{Var}(Y)}$.*

$h(X)$: Since $\text{Var}(X)=1$, the initial differential entropy is:


$h(X) = \frac{1}{2}\ln(2\pi e (1)) = \frac{1}{2}\ln(2\pi e) \text{ nats}$

$h(X \mid Y)$: computing the conditional variance $\sigma_{X|Y}^2 = \sigma_X^2 - \frac{\text{Cov}(X,Y)^2}{\text{Var}(Y)}$.


$\sigma_{X|Y}^2 = 1 - \frac{1^2}{5} = 1 - 0.2 = 0.8$


The conditional differential entropy is:


$h(X \mid Y) = \frac{1}{2}\ln(2\pi e (0.8)) \text{ nats}$

$h(X \mid Y) = 1.886  \text { bits}$

**6. Mutual Information**
Calculate $I(X; Y) = h(X) - h(X \mid Y)$.
* What percentage of the "uncertainty" about the temperature is removed by observing the power meter?
* How would this change if the mechanical noise $Q$ had a variance of $100$ instead of $4$?

$I(X;Y) = h(X) - h(X \mid Y) = \frac{1}{2}\ln(2\pi e) - \frac{1}{2}\ln(2\pi e \cdot 0.8) = \frac{1}{2}\ln\left(\frac{1}{0.8}\right) = \frac{1}{2}\ln(1.25) \approx 0.111 \text{ nats}$

Percentage of uncertainty removed:  $\frac{I(X;Y)}{h(X)} \approx \frac{0.161}{2.047} \approx \mathbf{7.87\%}$.

If mechanical noise ($Q$) had a variance of 100: The total variance of $Y$ would become $1 + 100 = 101$. The conditional variance would be $1 - \frac{1}{101} = \frac{100}{101} \approx 0.99$. Because the mechanical noise drastically overpowers the thermal signal, observing the power meter would only remove roughly $1\%$ of the uncertainty.

**7. Diagnostic Interpretation**
The factory manager says: *"Since $Y = X + Q$, our best estimate of temperature should just be $Y$ itself."*
Using your result from Question 3, explain why $Y$ is a **biased** estimator of $X$ and why $E[X \mid Y]$ "shrinks" the measurement toward the mean (zero).


If the manager simply uses $Y$ to estimate $X$, they are subjected to an expected estimation error (MSE) of $\text{Var}(X - Y) = \text{Var}(-Q) = 4$. This is a highly unoptimized approach because it treats all mechanical noise as if it were a temperature change.

The mathematically optimal estimator is $E[X \mid Y] = \frac{1}{5}Y$, which yields the absolute minimum possible error (an MSE of 0.8) works as an elegant balance between prior knowledge and new evidence. Because the correlation between $X$ and $Y$ is relatively weak and the sensor is known to be noisy, the conditional expectation deliberately "shrinks" the measurement toward the historical mean of zero (using the $\frac{1}{5}$ multiplier). The algorithm mathematically understands that an extreme power reading ($Y$) is far more likely to be caused by a random spike in cutting torque ($Q$) than a massive failure in temperature ($X$), and it dampens the prediction accordingly.
