#Q3) Transformer Failures



In a certain electricity distribution district, the number of emergency transformer failures reported in a single day is modeled as a discrete random variable $X$. Assume that $X$ takes values in the set $\{0,1,2,3\}$, with probabilities

$$
P(X=0)=0.50,\qquad P(X=1)=0.30,\qquad P(X=2)=0.15,\qquad P(X=3)=0.05.
$$

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.

$\Omega = \{0, 1, 2, 3\}$


$\mathfrak{F} = 2^\Omega$.

 $2^4 = 16$

$\mathfrak{F} = \{ \emptyset, \{0\}, \{1\}, \{2\}, \{3\}, \{0,1\}, \{0,2\}, \{0,3\}, \{1,2\}, \{1,3\}, \{2,3\}, \{0,1,2\}, \{0,1,3\}, \{0,2,3\}, \{1,2,3\}, \{0,1,2,3\}$

$P(\{0\}) = 0.50$

$P(\{1\}) = 0.30$

$P(\{2\}) = 0.15$

$P(\{3\}) = 0.05$

$P(\emptyset) = 0$

$P(\Omega) = P(\{0,1,2,3\}) = 1$

2. Define $X$ as a random variable on this space.

$X(\omega) = \omega \quad \text{for all } \omega \in \Omega$

3. Determine the probability mass function of $X$.

$p_X(x) = \begin{cases}
0.50 & \text{if } x = 0 \\
0.30 & \text{if } x = 1 \\
0.15 & \text{if } x = 2 \\
0.05 & \text{if } x = 3 \\
0 & \text{otherwise}
\end{cases}$

4. Compute the expectation $E[X]$.

$E[X] = \sum x \cdot P(X = x)$

$E[X] = (0)(0.50) + (1)(0.30) + (2)(0.15) + (3)(0.05)$

$E[X] = 0 + 0.30 + 0.30 + 0.15$

$E[X] = 0.75$

5. Compute the variance $\mathrm{Var}(X)$.

$E[X^2] = \sum x^2 \cdot P(X = x)$

$E[X^2] = (0^2)(0.50) + (1^2)(0.30) + (2^2)(0.15) + (3^2)(0.05)$

$E[X^2] = 0(0.50) + 1(0.30) + 4(0.15) + 9(0.05)$

$E[X^2] = 0 + 0.30 + 0.60 + 0.45 = 1.35$

$\text{Var}(X) = E[X^2] - (E[X])^2$

$\text{Var}(X) = 1.35 - (0.75)^2$

$\text{Var}(X) = 0.7875$

6. Find $P(X\ge 1)$ and $P(X\ge 2)$.

$P(X \ge 1) = 1 - P(X = 0) = 1 - 0.50 = 0.50$

$P(X \ge 2) = P(X = 2) + P(X = 3)$

$P(X \ge 2) = 0.15 + 0.05 = 0.20$

7. Briefly explain the practical meaning of the expectation and variance in this setting.

Expectation ($E[X] = 0.75$) conveys that the electricity distribution district can expect an average of 0.75 trensformer failures per day.

Variance ($\text{Var}(X) = 0.7875$) indicates that the actual number of failures per day will fluctuate by a value of 0.7875 from the average failures per day.

$$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x^{1}(1-x)^{17}, & 0<x<1,\\\
0, & \text{otherwise},
\end{cases}
$$

where $B(2,18)$ is the Beta function.

Answer the following:

1. Construct a probability space $(\Omega,\mathfrak F,P)$ suitable for this model.

$\Omega = [0, 1]$

$\mathfrak{F} = \sigma(\{ (a, b) \mid 0 \le a < b \le 1 \})$

$P(A) = \int_A \frac{1}{B(2, 18)} x^1 (1 - x)^{17} \, dx$

$X(\omega) = \omega \quad \text{for } \omega \in \Omega$

$
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x^{1}(1-x)^{17}, & 0<x<1,\\\
0, & \text{otherwise},
\end{cases}
$

4. Verify that $f$ is a valid probability density function.

A valid PDF must be non-negative and its total integral over the domain must equal 1.

 For any $x \in (0, 1)$, both $x$ and $(1-x)$ are strictly positive, so $f(x) \ge 0$.

 $\int_0^1 \frac{1}{B(2,18)} x^{2-1}(1-x)^{18-1} \,dx = \frac{B(2,18)}{B(2,18)} = 1$

$E[X] = \frac{\alpha}{\alpha + \beta}$

$\text{Var}(X) = \frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)}$

$\text{Var}(X) = \frac{2 \times 18}{(20)^2(21)} $

$$

$$

You may leave your answer in Beta integral form.

$P(X \le 0.10) = \int_{0}^{0.10} \frac{1}{B(2, 18)} x(1 - x)^{17} \,dx$

8. Find the probability that the defective proportion lies between $0.05$ and $0.15$, that is, compute.

$P(0.05 \le X \le 0.15) = \int_{0.05}^{0.15} \frac{1}{B(2, 18)} x(1 - x)^{17} \,dx$

It naturally models proportions (since it is bounded between 0 and 1), and its shape parameters easily represent a low, stable defect rate clustered near zero.

# Q5) Expected Project Completion Time

A construction company is estimating the time required to install a backup generator at a hospital. Let $X$ denote the number of hours needed to complete the installation. Based on expert judgment, the shortest possible time is $4$ hours, the longest possible time is $10$ hours, and the most likely completion time is $6$ hours. Assume that $X$ follows a Triangular distribution with parameters $a=4$, $b=10$, and $c=6$.

Answer the following:

1. Specify the support of $X$.

$X=[4,10]$

2. Write down the probability density function $f_X(x)$.

$f_X(x) = \begin{cases}
\frac{x - 4}{6} & \text{for } 4 \le x \le 6 \\
\frac{10 - x}{12} & \text{for } 6 < x \le 10 \\
0 & \text{otherwise}
\end{cases}$

3. Verify that $f_X$ integrates to $1$ over its support.

Base of the triangle: $b - a = 10 - 4 = 6$

Height of the triangle(at the peak $c=6$): $\frac{2}{b - a} = \frac{2}{6} = \frac{1}{3}$

Area under the triangle;

$=\frac{1}{2} \times \text{Base} \times \text{Height}$

$ = \frac{1}{2} \times 6 \times \frac{1}{3}$

$ = 1$


4. Compute the expectation $E[X]$.

$E[X] = \frac{a + b + c}{3} $

$E[X] = \frac{4 + 10 + 6}{3}$

$ E[X] \approx {6.67 \text{ hours}}$

$\text{Var}(X) \approx {1.56}$

6. Interpret the values of $a$, $b$, and $c$ in the context of this problem.

$a = 4$ :The best case scenario

$b = 10$ : The worst case scenario

A triangular distribution is a great choice as it perfectly matches how human experts naturally guess. When a construction project lacks enough past data to use complex statistics, we only need an expert to provide three simple estimates: the best-case, worst-case, and most likely times. This model easily turns those three basic, intuitive numbers into a complete and highly useful mathematical tool for planning.

#Q6) Expected Project Completion Time Using PERT

A project manager is estimating the time required to complete a critical software integration task. Let $X$ denote the number of days required to complete the task. Based on expert judgment, the shortest possible time is $a=8$ days, the longest possible time is $b=20$ days, and the most likely completion time is $c=11$ days. Assume that $X$ follows a PERT distribution with parameters $(a,b,c)$.

The PERT distribution is defined as a Beta distribution scaled to the interval $[a,b]$, with internal shape parameters

$$
\alpha = 1 + 4\left( \frac{c-a}{b-a} \right)
$$

and

$$
\beta = 1 + 4\left( \frac{b-c}{b-a} \right).
$$

Its probability density function is

$$
f(x)=\frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}{B(\alpha,\beta)(b-a)^{\alpha+\beta-1}}, \qquad a \leq x \leq b,
$$

where

$$
B(\alpha,\beta)=\int_0^1 t^{\alpha-1}(1-t)^{\beta-1},dt.
$$

Answer the following:

1. Compute the values of the shape parameters $\alpha$ and $\beta$.



$\alpha = 1 + 4 \left(\frac{c - a}{b - a}\right) = 1 + 4 \left(\frac{11 - 8}{20 - 8}\right) = 2$

$\beta = 1 + 4 \left(\frac{b - c}{b - a}\right) = 1 + 4 \left(\frac{20 - 11}{20 - 8}\right)= 4$

2. Specify the support of $X$.

$X=[8,20]$

3. Write down the probability density function $f_X(x)$ for this model.

$f_X(x) = \begin{cases}
\frac{(x - 8)^1 (20 - x)^3}{B(2, 4) (12)^5} & \text{for } 8 \le x \le 20 \\
0 & \text{otherwise}
\end{cases}$

4. State the probability space $(\Omega,\mathfrak F,P)$ suitable for this continuous model.

5. Define $X$ as a random variable on this space.

6. Show that the expectation $E[X]$ is given by

$$
E[X]=\frac{a+4c+b}{6}.
$$

$E[X] = \frac{a + 4c + b}{6}$

$E[X] = \frac{8 + 4(11) + 20}{6} = {12 \text{ days}}$

7. Compute the variance $\mathrm{Var}(X)$ using the approximation

$$
\mathrm{Var}(X)=\frac{(E[X]-a)(b-E[X])}{7}.
$$


$\text{Var}(X) = \frac{(E[X] - a)(b - E[X])}{7}$

$\text{Var}(X) = \frac{(12 - 8)(20 - 12)}{7}{\approx 4.57}$

8. Compute the approximate standard deviation $\sigma$.

$\sigma = \sqrt{\frac{32}{7}} \approx \mathbf{2.14 \text{ days}}$

9. Briefly explain why a PERT distribution may be more appropriate than a Triangular distribution in this setting byt comparing the two results.

A PERT distribution is more appropriate because it forms a smooth, bell-like curve that heavily concentrates probability around the most likely estimate ($c=11$), accurately reflecting how human expert estimates tend to cluster in project planning, rather than a Triangular distribution which spreads probability more heavily into the extreme best or worst-case tails.

10. Interpret the practical meaning of the values $a$, $b$, $c$, $E[X]$, and $\sigma$ in the context of project planning.

These numbers provide a practical schedule for the project. The fastest time is 8 days, the worst delay is 20 days, and the most common guess is 11 days. Since the worst-case scenario is so long, it pulls the safe average up to 12 days, which is the best number to use for official budgeting. The standard deviation gives us about 2 days of wiggle room around that 12-day target.
