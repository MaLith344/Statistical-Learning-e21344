# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

#1. The Partition Function


We begin with the definition of the canonical partition function and substitute the given energy eigenvalues:


$$Z(T) = \sum_{j=0}^{\infty} e^{-\frac{\hbar\omega(j + 1/2)}{k_B T}}$$

Using the properties of exponents, we can split the term inside the summation:


$$Z(T) = \sum_{j=0}^{\infty} \left( e^{-\frac{\hbar\omega j}{k_B T}} \cdot e^{-\frac{\hbar\omega}{2k_B T}} \right)$$

Because the term $e^{-\frac{\hbar\omega}{2k_B T}}$ does not depend on the summation index $j$, it acts as a constant and can be factored out of the sum:


$$Z(T) = e^{-\frac{\hbar\omega}{2k_B T}} \sum_{j=0}^{\infty} \left( e^{-\frac{\hbar\omega}{k_B T}} \right)^j$$

The remaining infinite sum is a geometric series of the form $\sum_{j=0}^{\infty} r^j$, where the common ratio is $r = e^{-\frac{\hbar\omega}{k_B T}}$. Since $T > 0$ and all constants are positive, we know that $r < 1$. Therefore, the series converges to $\frac{1}{1 - r}$.

Substituting the sum of the series back into our expression yields:


$$Z(T) = e^{-\frac{\hbar\omega}{2k_B T}} \left( \frac{1}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \right) = \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}}$$

#2. Thermodynamic Potentials


Helmholtz Free Energy, $F(T)$:
Using the thermodynamic relation $F(T) = -k_B T \ln Z$:


$$F(T) = -k_B T \ln \left( \frac{e^{-\frac{\hbar\omega}{2k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \right)$$

Applying the logarithmic quotient rule $\ln(A/B) = \ln A - \ln B$:


$$F(T) = -k_B T \left[ \ln\left(e^{-\frac{\hbar\omega}{2k_B T}}\right) - \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right]$$

Simplifying the natural log of the exponential:


$$F(T) = -k_B T \left[ -\frac{\hbar\omega}{2k_B T} - \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right]$$

Distributing the $-k_B T$ gives the final expression for $F(T)$:


$$F(T) = \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right)$$

Internal Energy, $U(T)$:
Using the statistical definition $U(T) = -\frac{\partial \ln Z}{\partial \beta}$, where $\beta = \frac{1}{k_B T}$:
First, let's express $\ln Z$ strictly in terms of $\beta$:


$$\ln Z = -\frac{\beta\hbar\omega}{2} - \ln(1 - e^{-\beta\hbar\omega})$$

Now, take the partial derivative with respect to $\beta$:


$$\frac{\partial \ln Z}{\partial \beta} = -\frac{\hbar\omega}{2} - \left( \frac{1}{1 - e^{-\beta\hbar\omega}} \right) \cdot (\hbar\omega e^{-\beta\hbar\omega})$$

$$\frac{\partial \ln Z}{\partial \beta} = -\frac{\hbar\omega}{2} - \frac{\hbar\omega e^{-\beta\hbar\omega}}{1 - e^{-\beta\hbar\omega}}$$

Since $U(T) = -\frac{\partial \ln Z}{\partial \beta}$:


$$U(T) = \frac{\hbar\omega}{2} + \frac{\hbar\omega e^{-\beta\hbar\omega}}{1 - e^{-\beta\hbar\omega}}$$

To clean this up, multiply the numerator and denominator of the second term by $e^{\beta\hbar\omega}$:


$$U(T) = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\beta\hbar\omega} - 1} = \hbar\omega \left( \frac{1}{2} + \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \right)$$

#3. Entropy


Finding $S(T)$:


$$S(T) = -\frac{\partial}{\partial T} \left[ \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right]$$

The first term is a constant with respect to $T$, so its derivative is zero. For the second term, we apply the product rule:


$$S(T) = -\left[ k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + k_B T \frac{\partial}{\partial T} \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right]$$

Let's calculate the derivative part via the chain rule:


$$\frac{\partial}{\partial T} \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) = \left(\frac{1}{1 - e^{-\frac{\hbar\omega}{k_B T}}}\right) \cdot \left(-e^{-\frac{\hbar\omega}{k_B T}}\right) \cdot \left(\frac{\hbar\omega}{k_B T^2}\right)$$

Substituting this back into the entropy equation:


$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) - k_B T \left( \frac{-e^{-\frac{\hbar\omega}{k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \right) \left(\frac{\hbar\omega}{k_B T^2}\right)$$

$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) - \frac{\hbar\omega}{T} \left( \frac{-e^{-\frac{\hbar\omega}{k_B T}}}{1 - e^{-\frac{\hbar\omega}{k_B T}}} \right)$$

Multiplying the numerator and denominator of the fraction by $e^{\frac{\hbar\omega}{k_B T}}$ simplifies it:


$$S(T) = -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T} \left( \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \right)$$

Verifying $U = F + TS$:
Substitute the explicit formulas for $F(T)$ and $S(T)$:


$$F + TS = \left[ \frac{\hbar\omega}{2} + k_B T \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) \right] + T \left[ -k_B \ln\left(1 - e^{-\frac{\hbar\omega}{k_B T}}\right) + \frac{\hbar\omega}{T} \left( \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \right) \right]$$

Distribute the $T$ in the second bracket:


$$F + TS = \frac{\hbar\omega}{2} + k_B T \ln\left(1 - \dots\right) - k_B T \ln\left(1 - \dots\right) + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1}$$

The logarithmic terms perfectly cancel each other out, leaving:


$$F + TS = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\frac{\hbar\omega}{k_B T}} - 1} = \hbar\omega \left( \frac{1}{2} + \frac{1}{e^{\frac{\hbar\omega}{k_B T}} - 1} \right)$$


This exactly equals our previously derived $U(T)$, verifying the consistency of the thermodynamic identities.

#4. Heat Capacity and Hyperbolic Form


Calculating $C_V$:
Start with $C_V = \left(\frac{\partial U}{\partial T}\right)$:


$$C_V = \frac{\partial}{\partial T} \left[ \frac{\hbar\omega}{2} + \hbar\omega \left( e^{\frac{\hbar\omega}{k_B T}} - 1 \right)^{-1} \right]$$

Applying the chain rule:


$$C_V = \hbar\omega \cdot (-1)\left( e^{\frac{\hbar\omega}{k_B T}} - 1 \right)^{-2} \cdot \left( e^{\frac{\hbar\omega}{k_B T}} \right) \cdot \left( -\frac{\hbar\omega}{k_B T^2} \right)$$

$$C_V = k_B \left( \frac{\hbar\omega}{k_B T} \right)^2 \frac{e^{\frac{\hbar\omega}{k_B T}}}{\left( e^{\frac{\hbar\omega}{k_B T}} - 1 \right)^2}$$

Conversion to Hyperbolic Form:
Let's analyze the fractional part. We can divide both the numerator and the denominator by $e^{\frac{\hbar\omega}{k_B T}}$:


$$\frac{1}{e^{-\frac{\hbar\omega}{k_B T}}\left( e^{\frac{\hbar\omega}{k_B T}} - 1 \right)^2} = \frac{1}{\left( e^{\frac{\hbar\omega}{2k_B T}} - e^{-\frac{\hbar\omega}{2k_B T}} \right)^2}$$

Recall the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$, which implies $2\sinh(x) = e^x - e^{-x}$.
Squaring both sides gives $(2\sinh(x))^2 = (e^x - e^{-x})^2$.

By setting $x = \frac{\hbar\omega}{2k_B T}$, the denominator matches this perfectly:


$$\left( e^{\frac{\hbar\omega}{2k_B T}} - e^{-\frac{\hbar\omega}{2k_B T}} \right)^2 = 4\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)$$

Substituting this back into the $C_V$ equation yields:


$$C_V = k_B \left( \frac{\hbar\omega}{k_B T} \right)^2 \frac{1}{4\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}$$

Absorbing the $\frac{1}{4}$ into the squared term gives the final elegant form:


$$C_V = k_B \left( \frac{\hbar\omega}{2k_B T} \right)^2 \frac{1}{\sinh^2\left(\frac{\hbar\omega}{2k_B T}\right)}$$

#5. Asymptotic Behavior


In the high-temperature limit, thermal energy is significantly greater than the quantum energy spacing, meaning $k_B T \gg \hbar\omega$.

Let's define $x = \frac{\hbar\omega}{2k_B T}$. In this high-temperature limit, $x$ becomes infinitesimally small ($x \to 0$).
For very small values of $x$, we can use the Taylor expansion for the hyperbolic sine function, which to first-order is simply $\sinh(x) \approx x$.

Substituting this approximation into our hyperbolic form of the heat capacity:


$$C_V = k_B (x)^2 \frac{1}{\sinh^2(x)} \approx k_B \frac{x^2}{x^2} = k_B$$

In the high-temperature limit, the quantum heat capacity $C_V$ asymptotically converges to the constant $k_B$.

This is in perfect agreement with the classical equipartition theorem. According to classical statistical mechanics, every independent quadratic degree of freedom contributes $\frac{1}{2}k_B$ to the heat capacity of a system. A 1-dimensional classical harmonic oscillator has exactly two such degrees of freedom:

Kinetic energy (proportional to momentum squared, $p^2$)

Potential energy (proportional to position squared, $q^2$)

Since $2 \times \frac{1}{2}k_B = k_B$, the total theoretical classical heat capacity is $k_B$. Our quantum mechanical derivation converges exactly to this classical limit when thermal fluctuations "wash out" the discrete quantum nature of the energy levels.

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$


We seek to maximize the continuous Shannon entropy functional (assuming a uniform reference measure):


$$H[p] = -\int_{-\infty}^{\infty} p(x) \ln p(x) dx$$


To enforce the constraints during optimization, we use the calculus of variations and introduce three Lagrange multipliers ($\lambda_0, \lambda_1, \lambda_2$):


$$\mathcal{L}[p] = -\int_{-\infty}^{\infty} p(x) \ln p(x) dx - \lambda_0 \left( \int_{-\infty}^{\infty} p(x) dx - 1 \right) - \lambda_1 \left( \int_{-\infty}^{\infty} x p(x) dx - \mu \right) - \lambda_2 \left( \int_{-\infty}^{\infty} (x - \mu)^2 p(x) dx - \sigma^2 \right)$$


For $\mathcal{L}[p]$ to be stationary (at a maximum), the functional derivative with respect to $p(x)$ must be exactly zero:


$$\frac{\delta \mathcal{L}}{\delta p(x)} = -\ln p(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0$$

Solving algebraically for the probability density $p(x)$:


$$p(x) = \exp\left( -1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2 \right)$$

We can define a normalization constant (partition function) $Z = \exp(1 + \lambda_0)$ to absorb the first two constant terms. This yields the general form:


$$p(x) = \frac{1}{Z} \exp\left( -\lambda_1 x - \lambda_2 (x - \mu)^2 \right)$$


We now apply the original constraints to solve for $\lambda_1, \lambda_2,$ and $Z$.

Applying the Mean Constraint ($\mathbb{E}[X] = \mu$):
For the expected value of $X$ to equal $\mu$, the distribution must be symmetrically centered around $\mu$. In the exponential argument $-\lambda_1 x - \lambda_2 (x - \mu)^2$, the presence of a non-zero $\lambda_1$ would shift the mean away from $\mu$. Therefore, the linear coefficient must vanish:


$$\lambda_1 = 0$$


The distribution simplifies to:


$$p(x) = \frac{1}{Z} \exp\left( -\lambda_2 (x - \mu)^2 \right)$$

Applying Variance ($\sigma^2$) and Normalization (1) Constraints:
We must satisfy:


$$1 = \int_{-\infty}^{\infty} \frac{1}{Z} \exp\left( -\lambda_2 (x - \mu)^2 \right) dx$$

$$\sigma^2 = \int_{-\infty}^{\infty} \frac{(x - \mu)^2}{Z} \exp\left( -\lambda_2 (x - \mu)^2 \right) dx$$


These are standard Gaussian integrals. By evaluating them (using the known standard forms $\int e^{-ay^2}dy = \sqrt{\pi/a}$ and $\int y^2 e^{-ay^2}dy = \frac{1}{2a}\sqrt{\pi/a}$), we find that to satisfy a variance of $\sigma^2$:


$$\lambda_2 = \frac{1}{2\sigma^2}$$


Substituting $\lambda_2$ into the normalization integral allows us to solve for $Z$:


$$Z = \sqrt{2\pi\sigma^2}$$


Substituting the resolved constants ($\lambda_1 = 0$, $\lambda_2 = \frac{1}{2\sigma^2}$, $Z = \sqrt{2\pi\sigma^2}$) back into our optimized density function $p(x)$:

$$p^*(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

Conclusion: The derived probability density function is identically the Normal (Gaussian) distribution. Therefore, the Gaussian distribution maximizes entropy for a given mean and variance. (Q.E.D.)

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform automatically routes inbound customer support emails to one of four mutually exclusive target destinations ($Y$):

* $Y = 0$ : **Finance**
* $Y = 1$ : **Technical**

Before final classification occurs, an upstream **AI Feature Extractor Agent** processes the raw text of each incoming email and outputs a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T \in \mathscr{X}$, where:

* $x_1 = \mathbb{I}(\text{Financial context detected})$ (e.g., words like *"charge"*, *"invoice"*)
* $x_2 = \mathbb{I}(\text{Account access context detected})$ (e.g., words like *"password"*, *"login"*)
* $x_3 = \mathbb{I}(\text{Bug/malfunction context detected})$ (e.g., words like *"error"*, *"frozen"*)

---

Because human language inherently overlaps, the AI Agent's outputs are distributed non-deterministically across three common operational scenarios. The marginal probability of observing each state is fixed by historical logs to:

1. **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies P(\mathbf{x}_A) = 0.40$
2. **Scenario B (Pure Acces):** $\mathbf{x}_B = (0, 1, 0)^T \implies P(\mathbf{x}_B) = 0.25$
3. **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies P(\mathbf{x}_C) = 0.05$
3. **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies P(\mathbf{x}_D) = 0.15$
3. **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies P(\mathbf{x}_E) = 0.05$

---

Historical manual audits provide the following observed joint dataset expectations:

* Constraint 1: $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = \mathbf{0.05}$
(Very low. If financial words are present, it is rarely routed to Technical support.)

* Constraint 2: $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = \mathbf{0.15}$
(Moderate. Account access issues split down the middle depending on context.)

* Constraint 3: $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = \mathbf{0.16}$
(High. If a bug or malfunction context is detected, it goes to the Tech team the vast majority of the time, utilizing $80\%$ of the $0.20$ max capacity).





According to Jaynes' Principle of Maximum Entropy, the least-biased probability distribution is the one that maximizes the conditional Shannon entropy $H(Y \mid \mathbf{x})$ subject to the known empirical constraints.

For a binary classification problem ($Y \in \{0, 1\}$), the maximum entropy distribution mathematically collapses into the standard sigmoid (logistic) function. Using a weight vector $\mathbf{w} = (w_1, w_2, w_3)^T$ for the three features, the probability that an email is routed to Technical Support ($Y = 1$) given an input state $\mathbf{x}$ is formulated as:

$$P(Y = 1 \mid \mathbf{x}) = \frac{1}{1 + e^{-\mathbf{w}^T \mathbf{x}}} = \sigma(w_1 x_1 + w_2 x_2 + w_3 x_3)$$


The historical audits provide macroscopic constraints. We must match the expected value of our model's predictions to these real-world averages using the formula:

$$\mathbb{E}[x_i \cdot \mathbb{I}(Y = 1)] = \sum_{\mathbf{x}} x_i P(Y = 1 \mid \mathbf{x}) P(\mathbf{x})$$

By expanding this expectation across the five common scenarios ($A, B, C, D, E$), we map the active features to their respective scenarios:

Constraint 1 (Financial context, $x_1$): Active only in Scenarios A and E.


$$P(\mathbf{x}_A)P(Y = 1 \mid \mathbf{x}_A) + P(\mathbf{x}_E)P(Y = 1 \mid \mathbf{x}_E) = 0.05$$

$$0.40 \cdot \sigma(w_1) + 0.05 \cdot \sigma(w_1 + w_2) = 0.05$$

Constraint 2 (Account access context, $x_2$): Active only in Scenarios B, D, and E.


$$P(\mathbf{x}_B)P(Y = 1 \mid \mathbf{x}_B) + P(\mathbf{x}_D)P(Y = 1 \mid \mathbf{x}_D) + P(\mathbf{x}_E)P(Y = 1 \mid \mathbf{x}_E) = 0.15$$

$$0.25 \cdot \sigma(w_2) + 0.15 \cdot \sigma(w_2 + w_3) + 0.05 \cdot \sigma(w_1 + w_2) = 0.15$$

Constraint 3 (Bug/malfunction context, $x_3$): Active only in Scenarios C and D.


$$P(\mathbf{x}_C)P(Y = 1 \mid \mathbf{x}_C) + P(\mathbf{x}_D)P(Y = 1 \mid \mathbf{x}_D) = 0.16$$

$$0.05 \cdot \sigma(w_3) + 0.15 \cdot \sigma(w_2 + w_3) = 0.16$$


To find the exact parameters that satisfy these constraints, we solve the system of non-linear equations (which represents the gradient descent optimization seeking informational "equilibrium").

Using numerical root-finding, the optimal weights are determined to be:

$w_1 \approx -1.9725$ (Financial context strongly pulls the probability toward 0, i.e., Finance)

$w_2 \approx -1.7762$ (Access context pulls the probability toward 0, i.e., Finance)

$w_3 \approx 2.8804$ (Bug/Malfunction context strongly pushes the probability toward 1, i.e., Technical)

Step 4: Evaluate the Final Routing Probabilities
Using the learned weights $\mathbf{w}$, we can establish the operational routing logic for the upstream AI agent by plugging the dot products back into the sigmoid function:

Scenario A (Pure Financial): $\mathbf{x}_A = (1, 0, 0)^T$


$$P(Y = 1 \mid \mathbf{x}_A) = \sigma(-1.9725) \approx 12.21\%$$


(Mostly routed to Finance, as expected).

Scenario B (Pure Access): $\mathbf{x}_B = (0, 1, 0)^T$


$$P(Y = 1 \mid \mathbf{x}_B) = \sigma(-1.7762) \approx 14.48\%$$


(Generally routed to Finance when standing alone).

Scenario C (Pure Technical): $\mathbf{x}_C = (0, 0, 1)^T$


$$P(Y = 1 \mid \mathbf{x}_C) = \sigma(2.8804) \approx 94.69\%$$


(Almost exclusively routed to Technical).

Scenario D (Access Technical): $\mathbf{x}_D = (0, 1, 1)^T$


$$P(Y = 1 \mid \mathbf{x}_D) = \sigma(-1.7762 + 2.8804) = \sigma(1.1042) \approx 75.10\%$$

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.




According to the principles of continuous Maximum Entropy inference, we seek to construct a probability distribution that maximizes our uncertainty without introducing arbitrary bias beyond our known macroscopic measurements. Assuming a uniform baseline reference measure $m(S_T) = 1$ over the domain $S_T \in (0, \infty)$, the continuous Shannon-Jaynes entropy functional to be maximized is defined as:

$$H[p] = -\int_{0}^{\infty} p(S_T) \ln p(S_T) dS_T$$


Our optimization problem is strictly bounded by the four macroscopic constraints extracted from the market data:

Normalization: The total probability must sum to one:

$$\int_{0}^{\infty} p(S_T) dS_T = 1$$

The Forward Price: The expected value of the asset must match the risk-free growth:

$$\int_{0}^{\infty} S_T p(S_T) dS_T = 102$$

The Market Volatility: The expected squared price is constrained by:

$$\int_{0}^{\infty} S_T^2 p(S_T) dS_T = 10600$$

The Skewness/Crash Risk: The expected logarithmic penalty is constrained by:

$$\int_{0}^{\infty} \ln(S_T) p(S_T) dS_T = 4.60$$


Using the calculus of variations, we construct the functional Lagrangian $\mathcal{L}[p]$ by introducing distinct Lagrange multipliers ($\gamma, \lambda_1, \lambda_2, \lambda_3$) to enforce each of the respective constraints:

$$\mathcal{L}[p] = -\int_{0}^{\infty} p(S_T) \ln p(S_T) dS_T - \gamma\left(\int_{0}^{\infty} p(S_T) dS_T - 1\right) - \lambda_1\left(\int_{0}^{\infty} S_T p(S_T) dS_T - 102\right) - \lambda_2\left(\int_{0}^{\infty} S_T^2 p(S_T) dS_T - 10600\right) - \lambda_3\left(\int_{0}^{\infty} \ln(S_T) p(S_T) dS_T - 4.60\right)$$

To find the stationary point that maximizes the entropy, we take the functional derivative of $\mathcal{L}[p]$ with respect to the probability density $p(S_T)$ and set it exactly to zero:

$$\frac{\delta \mathcal{L}}{\delta p(S_T)} = -\ln p(S_T) - 1 - \gamma - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T) = 0$$

Solving this algebraic equation explicitly for $p(S_T)$ yields:

$$p^*(S_T) = \exp\left(-(1 + \gamma) - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln(S_T)\right)$$


By enforcing the normalization constraint, we can factor out the constant terms and eliminate the multiplier $\gamma$ by defining the continuous partition function $Z \triangleq \exp(1 + \gamma)$. Furthermore, applying the standard logarithmic identity $\exp(-\lambda_3 \ln(S_T)) = S_T^{-\lambda_3}$ allows us to simplify the expression into the unique, most unbiased continuous probability density function:

$$p^*(S_T) = \frac{1}{Z} S_T^{-\lambda_3} \exp\left(-\lambda_1 S_T - \lambda_2 S_T^2\right)$$


The derived PDF $p^*(S_T)$ represents a powerful, generalized exponential-family distribution that natively incorporates the asset's mean, variance, and asymmetric skewness.

Because this combination of $x, x^2$, and $\ln(x)$ constraints does not perfectly reduce to a standard elementary distribution (like a pure Gaussian or pure Lognormal), the exact numerical values of the Lagrange multipliers ($\lambda_1, \lambda_2, \lambda_3$) and the partition function $Z$ cannot be isolated analytically. In practice, a quantitative analyst solves for these parameters by substituting this functional form back into the integral constraint equations in Step 2 and utilizing a numerical root-finding optimization algorithm to find the exact distribution.
