# Q. Analytical Derivation

Consider the filter model
\begin{align*}
x^{-}_k &= A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}, \\
y^{-}_k &= H_k\,x^{-}_k + z_k,
\end{align*}
where
$$x^{+}_{k-1} \sim \mathscr{N}(m_{k-1}, P_{k-1}),$$

Answer the followings:

1. Show that
$$x_k^- \sim \mathscr{N}(m_k^-, P_k^-),$$
where
$$m_k^- = A_{k-1}m_{k-1},$$
and
$$
P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T.
$$

2. Show that
$$y_k^- \sim \mathscr{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m).$$

3. Show that
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right).$$

4. Show that
$$x^{+}_k\triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k),$$
where the updated mean $m_k$ and updated covariance $P_k$ are given by:
\begin{align*}
K_k &\triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}, \\
m_k &= m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-), \\
P_k &= (I - K_k H_k) P_k^-.
\end{align*}

5. Find
$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}]$  and
$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k})$.

# Q. Analytical Derivation (Answers)

### 1.
Given $m_k^- = A_{k-1}m_{k-1}$ and $P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T$:

Mean:$$\begin{aligned}
m_k^- &\triangleq E[x_k^-] \\
m_k^- &= E[A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1}] \\
m_k^- &= A_{k-1}E[x_{k-1}^+] + G_{k-1}E[w_{k-1}] \\
m_k^- &= A_{k-1}m_{k-1} + G_{k-1}(0) \\
m_k^- &= A_{k-1}m_{k-1}
\end{aligned}$$Variance:$$\begin{aligned}
P_k^- &\triangleq Var(x_k^-) \\
P_k^- &= Var(A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1}) \\
P_k^- &= A_{k-1}Var(x_{k-1}^+)A_{k-1}^T + G_{k-1}Var(w_{k-1})G_{k-1}^T \\
P_k^- &= A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T
\end{aligned}$$Resulting Distribution:$$x_k^- \sim \mathcal{N}(m_k^-, P_k^-)$$
### 2.  
Expected Value:$$\begin{aligned}
E[y_k^-] &= E[H_k x_k^- + z_k] \\
E[y_k^-] &= H_k E[x_k^-] + E[z_k] \\
E[y_k^-] &= H_k m_k^- + 0 \\
E[y_k^-] &= H_k m_k^-
\end{aligned}$$Variance:$$\begin{aligned}
Var(y_k^-) &= Var(H_k x_k^- + z_k) \\
Var(y_k^-) &= H_k Var(x_k^-)H_k^T + Var(z_k) \\
Var(y_k^-) &= H_k P_k^- H_k^T + \Sigma_m
\end{aligned}$$Resulting Distribution:$$y_k^- \sim \mathcal{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m)$$

### 3.

Covariance:$$\begin{aligned}
Cov(x_k^-, y_k^-) &= Cov(x_k^-, H_k x_k^- + z_k) \\
Cov(x_k^-, y_k^-) &= Cov(x_k^-, H_k x_k^-) + Cov(x_k^-, z_k) \\
Cov(x_k^-, y_k^-) &= P_k^- H_k^T + 0 \\
Cov(x_k^-, y_k^-) &= P_k^- H_k^T
\end{aligned}$$

Combining this cross-covariance with the individual means and variances derived above forms a block-structured multivariate Gaussian joint distribution:

$$\begin{bmatrix} x_k^- \\ y_k^- \end{bmatrix} \sim \mathcal{N} \left( \begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix}, \begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix} \right)$$

### 4.
When a physical measurement is observed ($y_k = y_k^{obs}$), the standard conditioning properties of multivariate normal distributions are applied to update the prior belief by slicing the joint distribution at the realization $y_k^{obs}$. This yields the posterior state distribution:

$$x_k^+ \triangleq (x_k^- \mid y_k^- = y_k^{obs}) \sim \mathcal{N}(m_k, P_k)$$

The extracted mean and covariance equations are derived as:
* Kalman Gain: $K_k \triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$
* Updated Mean: $m_k = m_k^- + K_k (y_k^{obs} - H_k m_k^-)$
* Updated Covariance: $P_k = (I - K_k H_k) P_k^-$

### 5.
Based on the posterior distribution extraction:
* The conditional expectation $E[x_k^- \mid y_k^- = y_k^{obs}]$ acts as the updated state estimate, $m_k$.
* The conditional variance $Var(x_k^- \mid y_k^- = y_k^{obs})$ is exactly the new posterior error covariance matrix, $P_k$.

# Q. 1-D Example

Consider the scalar linear-Gaussian filter model
\begin{aligned}
x^-_k &= a\,x^+_{k-1} + w_{k-1},\qquad w_{k-1}\sim\mathscr N(0,q),\\
y^-_k &= h\,x^-_k + z_k,\qquad\;\;\;\; z_k\sim\mathscr N(0,r),
\end{aligned}
with prior $x_0\sim \mathscr{N}(m_0,P_0)$. Define $Y_{k}=\{y^{\mathrm{obs}}_1,\dots,y^{\mathrm{obs}}_{k}\}$.

Answer the following:

1. Show that
\begin{aligned}
m_k^- &= a\,m_{k-1},\\
P_k^- &= a^2 P_{k-1} + q.
\end{aligned}

2. Show that
\begin{aligned}
m_k &= m_k^- + K_k\,v_k
= m_k^- + \frac{P_k^- h}{S_k}\bigl(y^{\mathrm{obs}}_k - h\,m_k^-\bigr),\\
P_k &= (1 - K_k h)\,P_k^-
= \Bigl(1 - \frac{P_k^- h^2}{S_k}\Bigr) P_k^- .
\end{aligned}

3. Show that the predictive measurement distribution (before seeing $y_k$):
\begin{align*}
p(y^-_k | Y_{k-1})=\mathscr N\bigl(h\,m_k^-,\; h^2 P_k^- + r\bigr).
\end{align*}

4. Show that the posterior-predictive measurement distribution (after filtering on $y_k$):
\begin{align*}
p(y^-_k\mid Y_k)=\mathscr N\bigl(h\,m_k,\; h^2 P_k + r\bigr).
\end{align*}

5. For suitable numerical values for $a,q,h,r$ animate the above to prior distribution and the posterior distribution.

1.

Mean:$$\begin{aligned}
m_k^- &= E[a x_{k-1}^+ + w_{k-1}] \\
m_k^- &= a E[x_{k-1}^+] + E[w_{k-1}] \\
m_k^- &= a m_{k-1} + 0 \\
m_k^- &= a m_{k-1}
\end{aligned}$$Variance:$$\begin{aligned}
P_k^- &= Var(a x_{k-1}^+ + w_{k-1}) \\
P_k^- &= a^2 Var(x_{k-1}^+) + Var(w_{k-1}) \\
P_k^- &= a^2 P_{k-1} + q
\end{aligned}$$

2.

Let the innovation (residual) be $v_k \triangleq y_k^{obs} - h m_k^-$ and the innovation covariance be $S_k$:$$\begin{aligned}
S_k &= Var(h x_k^- + z_k) = h^2 P_k^- + r \\
K_k &= \frac{P_k^- h}{S_k}
\end{aligned}$$Updating the Mean:$$\begin{aligned}
m_k &= m_k^- + K_k v_k \\
m_k &= m_k^- + \left(\frac{P_k^- h}{S_k}\right)(y_k^{obs} - h m_k^-)
\end{aligned}$$Updating the Variance:$$\begin{aligned}
P_k &= (1 - K_k h) P_k^- \\
P_k &= \left(1 - \left(\frac{P_k^- h}{S_k}\right)h\right) P_k^- \\
P_k &= \left(1 - \frac{P_k^- h^2}{S_k}\right) P_k^-
\end{aligned}$$

3.

$$$$\begin{aligned}
E[y_k^- \mid Y_{k-1}] &= h m_k^- \\
Var(y_k^- \mid Y_{k-1}) &= h^2 P_k^- + r
\end{aligned}
$$p(y_k^- \mid Y_{k-1}) = \mathcal{N}(h m_k^-, h^2 P_k^- + r)$$

4.

$$$$\begin{aligned}
E[y_k^- \mid Y_k] &= h m_k \\
Var(y_k^- \mid Y_k) &= h^2 P_k + r
\end{aligned}

$$p(y_k^- \mid Y_k) = \mathcal{N}(h m_k, h^2 P_k + r)$$

### 4. Animate the prior and posterior distributions


```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from scipy.stats import norm
from IPython.display import HTML

# Define parameters matching the 1-D example
a, q, h, r = 1.0, 0.5, 1.0, 1.0
m_prev, P_prev = 0.0, 1.0  # Initial state x0 ~ N(m0, P0)

# Simulated true state and observations
true_states = [0.0]
observations = []
np.random.seed(42)
for i in range(10):
    true_states.append(a * true_states[-1] + np.random.normal(0, np.sqrt(q)))
    observations.append(h * true_states[-1] + np.random.normal(0, np.sqrt(r)))

fig, ax = plt.subplots(figsize=(8, 5))
x_axis = np.linspace(-10, 10, 1000)
line_prior, = ax.plot([], [], 'r--', label='Prior p(x_k- | Y_{k-1})')
line_post, = ax.plot([], [], 'b-', label='Posterior p(x_k+ | Y_k)')
scatter_obs = ax.scatter([], [], color='k', zorder=5, label='Observation y_k')
ax.legend()
ax.set_xlim(-10, 10)
ax.set_ylim(0, 1.0)
ax.set_title("1-D Kalman Filter Animation")

def init():
    line_prior.set_data([], [])
    line_post.set_data([], [])
    scatter_obs.set_offsets(np.empty((0, 2)))
    return line_prior, line_post, scatter_obs

def update(frame):
    global m_prev, P_prev
    y_obs = observations[frame]

    # 1. Prediction (Prior)
    m_minus = a * m_prev
    P_minus = a**2 * P_prev + q

    # 2. Update (Posterior)
    S = h**2 * P_minus + r
    K = (P_minus * h) / S
    m_post = m_minus + K * (y_obs - h * m_minus)
    P_post = (1 - K * h) * P_minus

    # Update distributions for plotting
    line_prior.set_data(x_axis, norm.pdf(x_axis, m_minus, np.sqrt(P_minus)))
    line_post.set_data(x_axis, norm.pdf(x_axis, m_post, np.sqrt(P_post)))
    scatter_obs.set_offsets(np.array([[y_obs, 0]]))
    ax.set_title(f"Step {frame+1}: Prior to Posterior Update")

    # Prepare for next frame
    m_prev, P_prev = m_post, P_post
    return line_prior, line_post, scatter_obs

ani = animation.FuncAnimation(
    fig, update, frames=len(observations),
    init_func=init, blit=False, repeat=False  # <-- blit=False so the title actually updates
)

# Required to render animations smoothly in Google Colab
plt.close()
HTML(ani.to_jshtml())
```

# Q. 2D-Position Estimation

## Part A

Consider a two-dimensional constant-velocity dynamical system. The hidden state at time step $k$ is

$$
x_k =
\begin{bmatrix}
p_x(k)\\
p_y(k)\\
v_x(k)\\
v_y(k)
\end{bmatrix},
$$

where $(p_x(k),p_y(k))$ denote the position components and $(v_x(k),v_y(k))$ denote the velocity components.

The measurement consists only of the two position components:

$$
y_k =
\begin{bmatrix}
p_x^{\mathrm{meas}}(k)\\
p_y^{\mathrm{meas}}(k)
\end{bmatrix}.
$$

The linear Gaussian state-space filter model is

$$
x^-_k = A x^+_{k-1} + G w_{k-1},
$$

$$
y^-_k = Hx^+_k + z_k,
$$

where

$$
w_{k-1} \sim \mathscr{N}(0,\Sigma_p),
\qquad
z_k \sim \mathscr{N}(0,\Sigma_m).
$$

The process noise sequence $w_k$, measurement noise sequence $z_k$, and the initial state are assumed mutually independent.

Show that:

$$
A =
\begin{bmatrix}
1 & 0 & \Delta t & 0\\
0 & 1 & 0 & \Delta t\\
0 & 0 & 1 & 0\\
0 & 0 & 0 & 1
\end{bmatrix},
$$



$$
H =
\begin{bmatrix}
1 & 0 & 0 & 0\\
0 & 1 & 0 & 0
\end{bmatrix},
$$

and

$$
G =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0\\
0 & \frac{1}{2}\Delta t^2\\
\Delta t & 0\\
0 & \Delta t
\end{bmatrix}.
$$

# Q. 2D-Position Estimation (Answers)

### Part A
Assuming a sampling interval $\Delta t$, the discrete-time constant-velocity kinematic equations explicitly show that position is updated via velocity ($p(k) = p(k-1) + \Delta t v(k-1)$) and velocity remains constant ($v(k) = v(k-1)$). Arranging these equations into matrix form yields exactly the state transition matrix $A$:

Given the kinematic equations of motion:$$\begin{aligned}
p_x(k) &= p_x(k-1) + \Delta t v_x(k-1) \\
p_y(k) &= p_y(k-1) + \Delta t v_y(k-1) \\
v_x(k) &= v_x(k-1) \\
v_y(k) &= v_y(k-1)
\end{aligned}$$Expressing this in matrix form $x_k = A x_{k-1} + G w_{k-1}$:$$x_k = \begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix} =
\begin{bmatrix}
1 & 0 & \Delta t & 0 \\
0 & 1 & 0 & \Delta t \\
0 & 0 & 1 & 0 \\
0 & 0 & 0 & 1
\end{bmatrix}
\begin{bmatrix} p_x(k-1) \\ p_y(k-1) \\ v_x(k-1) \\ v_y(k-1) \end{bmatrix}
+ G w_{k-1}$$Therefore, the state-transition matrix $A$ is:$$A = \begin{bmatrix}
1 & 0 & \Delta t & 0 \\
0 & 1 & 0 & \Delta t \\
0 & 0 & 1 & 0 \\
0 & 0 & 0 & 1
\end{bmatrix}$$

Because the sensor measurement strictly isolates only the two position components ($y_k = [p_x(k), p_y(k)]^T + z_k$), the observation matrix $H$ is constructed to extract only the first two state variables:

Assuming we only observe the positions $p_x$ and $p_y$:$$y_k = \begin{bmatrix} p_x(k) \\ p_y(k) \end{bmatrix} + z_k$$Expressing this in matrix form $y_k = H x_k + z_k$:$$y_k = \begin{bmatrix}
1 & 0 & 0 & 0 \\
0 & 1 & 0 & 0
\end{bmatrix}
\begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix}
+ z_k$$Therefore, the observation matrix $H$ is:$$H = \begin{bmatrix}
1 & 0 & 0 & 0 \\
0 & 1 & 0 & 0
\end{bmatrix}$$

Unmodeled random acceleration over the sampling interval $\Delta t$ affects the position proportionally to $\frac{1}{2}\Delta t^2$ and affects the velocity proportionally to $\Delta t$. Collecting these integration factors yields the process-noise input matrix $G$:

Accounting for acceleration as the process noise ($w_{k-1}$):$$\begin{aligned}
p_x(k) &= p_x(k-1) + \Delta t v_x(k-1) + \frac{1}{2} \Delta t^2 a_x(k-1) \\
p_y(k) &= p_y(k-1) + \Delta t v_y(k-1) + \frac{1}{2} \Delta t^2 a_y(k-1) \\
v_x(k) &= v_x(k-1) + \Delta t a_x(k-1) \\
v_y(k) &= v_y(k-1) + \Delta t a_y(k-1)\\
\end{aligned}$$Extracting the coefficients for the acceleration inputs $a_x$ and $a_y$, we get the process-noise input matrix $G$:

$$G = \begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0 \\
0 & \frac{1}{2}\Delta t^2 \\
\Delta t & 0 \\
0 & \Delta t
\end{bmatrix}$$




## Part B

Develop a python based method to filter a sequence of noisy GPS obtained position meaurements.

```python
import numpy as np

class GPSKalmanFilter2D:
    def __init__(self, dt, q_variance, r_variance, initial_state, initial_cov):
        """
        Initializes the 2D Constant Velocity Kalman Filter.
        dt: Sampling interval
        q_variance: Intensity of the process noise (acceleration variance)
        r_variance: Measurement noise variance (GPS noise)
        """
        self.dt = dt

        # State Transition Matrix (A)
        # State vector: [x, y, vx, vy]
        self.A = np.array([
            [1, 0, dt,  0],
            [0, 1,  0, dt],
            [0, 0,  1,  0],
            [0, 0,  0,  1],
        ])

        # Observation Matrix (H) - we only observe position (x, y)
        self.H = np.array([
            [1, 0, 0, 0],
            [0, 1, 0, 0],
        ])

        # Process-Noise Input Matrix (G)
        self.G = np.array([
            [0.5 * dt**2, 0],
            [0, 0.5 * dt**2],
            [dt, 0],
            [0, dt]
        ])

        # Covariances
        self.Sigma_p = np.array([[q_variance, 0], [0, q_variance]])  # Acceleration noise
        self.Q = self.G @ self.Sigma_p @ self.G.T                    # State process covariance
        self.R = np.array([[r_variance, 0], [0, r_variance]])        # Measurement covariance (Sigma_m)

        # State and Uncertainty Initialization
        self.x = np.array(initial_state, dtype=float).reshape(4, 1)
        self.P = np.array(initial_cov, dtype=float)

    def predict(self):
        # Time Update (Prediction)
        self.x = self.A @ self.x
        self.P = self.A @ self.P @ self.A.T + self.Q
        return self.x.flatten()

    def update(self, measurement):
        # Measurement Update (Correction)
        y_obs = np.array(measurement).reshape(2, 1)

        # Innovation and Innovation Variance
        y_pred = self.H @ self.x
        S = self.H @ self.P @ self.H.T + self.R

        # Kalman Gain
        K = self.P @ self.H.T @ np.linalg.inv(S)

        # Update State and Covariance
        self.x = self.x + K @ (y_obs - y_pred)
        self.P = (np.eye(4) - K @ self.H) @ self.P

        return self.x.flatten()


# === Example Usage ===
# Simulate filtering a sequence of noisy GPS points (x, y)
gps_measurements = [(1.2, 1.1), (2.1, 2.3), (3.4, 2.9), (3.9, 4.2), (5.1, 4.8)]

# Initialize filter: dt=1s, low acceleration noise, high GPS noise
# Initial state [x, y, vx, vy] starting near the first measurement, zero initial velocity
kf = GPSKalmanFilter2D(
    dt=1.0, q_variance=0.1, r_variance=2.0,
    initial_state=[1.2, 1.1, 0, 0],
    initial_cov=np.eye(4) * 500
)

filtered_trajectory = []
for z in gps_measurements:
    kf.predict()
    state = kf.update(z)
    # Store filtered x and y (indices 0 and 1)
    filtered_trajectory.append((round(state[0], 2), round(state[1], 2)))

print("Original GPS Measurements:", gps_measurements)
print("Filtered GPS Trajectory (Positions):", filtered_trajectory)
```
